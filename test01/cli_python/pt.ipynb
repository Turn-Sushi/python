## 단어장 구현(python) 발표

### 0. 시작하며

- **데이터를 저장**하기 위해 ``JSON`` 파일을 사용함
- **명령어를 처리**하기 위해 ``argparse`` 라이브러리를 사용함

### 1. 데이터 구조 정의(?) → memo.json 파일 생성

"데이터가 저장될 공간인 ``memo.json`` 파일을 만들겠습니다."

- 데이터는 **words**라는 이름의 리스트 안에 쌓이게 되는 구조

````python
{
    "words" : []
}
````

### 2. 데이터 입출력 함수 만들기 → cmd.py 파일 생성

"데이터 파일(memo.json)을 **읽고(r) 쓰는(w) 기능**을 ```cmd.py```에 작성하겠습니다."

- import 및 FILE_PATH 작성
- getData, setData 작성


- ``getData()`` : 파일을 열어서 읽고 파이썬 딕셔너리(객체)로 변환해서 가져오는 역할
- ``setData()`` : 변경된 데이터를 다시 파일에 저장하는 역할
- ``list()`` : 저장 후 list를 바로 확인할 수 있게 list() 함수를 실행하게 함

### 3. 기능 구현(list) → cmd.py

"실제 기능을 추가하겠습니다. 단어 목록을 보여주는 ``list()`` 함수를 만들겠습니다."

- list() 함수는 데이터를 가져와 반복문으로 화면에 출력.
- 화면에 예쁘게 표 형태로 보여주기 위해 선을 만듦

### 4. 명령어 처리(list) → index.py 파일 생성

"사용자의 명령을 처리하기 위한 ``index.py``를 만들고 list명령어를 등록하겠습니다.

- ```argparse```를 사용해서 터미널에서 입력한 글자를 파이썬이 명령어로 인식하게 해줌

### 5. 기능 구현(insert) → cmd.py

"단어를 추가하는 ```insert()```를 만들겠습니다."

- 현재 번호(id) 중 가장 큰 값을 찾아 **+1**을 해줌으로써 중복되지 않는 번호를 만들 수 있음

### 6. 명령어 처리(insert) → index.py

### 8. 명령어 처리(update) → index.py
